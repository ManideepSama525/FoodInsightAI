# FoodInsightAI — Model Lab v1

This notebook is the reproducible starting point for the model decision. It does **not** assume Qwen, Llama, Gemma, or any other particular model.


## Experiment order
1. Inspect GPU/VRAM.
2. Run a small inference benchmark.
3. Measure memory and tokens/sec.
4. Compare candidate open-weight models.
5. Only then select the training target.
6. Build the FoodInsight dataset and run a small training pilot.


```python
import platform, sys, os
print('Python:', sys.version)
print('Platform:', platform.platform())
try:
    import torch
    print('PyTorch:', torch.__version__)
    print('CUDA:', torch.version.cuda)
    print('CUDA available:', torch.cuda.is_available())
    if torch.cuda.is_available():
        p = torch.cuda.get_device_properties(0)
        print('GPU:', p.name)
        print('VRAM GiB:', round(p.total_memory/1024**3, 2))
except Exception as exc:
    print('Torch unavailable:', repr(exc))
```

## Decision rule
Do not choose the final model by parameter count alone. Record latency, VRAM usage, answer quality, grounding/verification metrics, and training cost before selecting the model used in the final system.
